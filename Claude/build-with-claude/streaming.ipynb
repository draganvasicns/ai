```python
#Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
#Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role":"user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content": text}
    messages.append(assistant_message)


def stream_chat(messages, system=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
        "stream"     : True
    }

    if system :
        params["system"] = system


    stream = client.messages.create(**params)
    return stream

def stream_util(messages, system=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
    }

    if system :
        params["system"] = system


    stream = client.messages.stream(**params)
    return stream

def chat(messages, system=None, temperature=1.0):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
    }

    if system :
        params["system"] = system

   
    temperature_str = str(temperature)
    if("system" not in params):
        params["system"] = ""
    params["system"] = params["system"] + f"""If temperature: {temperature} 
    is low give precise, strict answer, but if temperature is high give creative answer. 
    Temperature goes from 0.0 to 1.0."""

    message = client.messages.create(**params);
        
    for msg in message.content:
        if(msg.type == "text"):
            return msg.text
```

```python
#Messages to send to claude
messages = []

#Create an prompt that could give long answer
add_user_message(messages, "What is quantum computing. Give me 5 sentences description.")


#send to claude, but in chat way
stream = stream_chat(messages)

#show result
for event in stream:
    print(event)
```
