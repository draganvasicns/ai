````python
#Load env variables
from dotenv import load_dotenv

load_dotenv()
````

````python
#Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
````

````python
def add_user_message(messages, text):
    user_message = {"role":"user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content": text}
    messages.append(assistant_message)


def stream_chat(messages, system=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
        "stream"     : True
    }

    if system :
        params["system"] = system


    stream = client.messages.create(**params)
    return stream

def stream_util(messages, system=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
    }

    if system :
        params["system"] = system


    stream = client.messages.stream(**params)
    return stream

def chat(messages, system=None, stop_sequences=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
    }

    if system :
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences
   
    message = client.messages.create(**params);
        
    for msg in message.content:
        if(msg.type == "text"):
            return msg.text
````

````python
#Messages to send to claude
messages = []

#Create an prompt that give json as answer
add_user_message(messages, """Generate a very short event bridge rule as json. 
                 Return just JSON without anything else.
                 Anwer should start with '{' and ends with '}' 
                 Example: { "source" ["a.b"]} """)

add_assistant_message(messages, "```json")
#send to claude, but in chat way
response = chat(messages, stop_sequences=["```"])

#show result
response
````
