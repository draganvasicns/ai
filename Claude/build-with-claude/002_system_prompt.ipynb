```python
#Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
#Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role":"user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model"      : model,
        "max_tokens" : 1000,
        "messages"   : messages,
    }

    if system is not None:
        params["system"] = system
    
    message = client.messages.create(**params);
        
    for msg in message.content:
        if(msg.type == "text"):
            return msg.text
```

```python
#Messages to send to claude
messages = []

#Create an equation with one unknown and ask claude to resolve it
add_user_message(messages, "How to resolve 2x+5=25?")

#Define system promt message
system_prompt = "You are a patient math tutor. Guide the user toward the answer instead of giving it immediately."

#send to claude
result = chat(messages, system_prompt)

#show result
result
```
